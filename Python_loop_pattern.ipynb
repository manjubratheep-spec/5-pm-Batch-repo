### **Python Loop patterns**
### ***For loop***

```python
for x in range(6):
    for y in range(1,6+1):
        print("*",end='')
    print()
```

```python
a = int(input("Enter the row:"))
symbol = input("Enter the symbol")
for i in range(a):
    for j in range(0,a+1):
        print(symbol,end='')
    print()
```

### ***While loop***

```python
a = 1
while a<=5:
    a += 1
    b=1
    while b<=5:
        b += 1
        print("*",end='')
    print()
```

```python
x= int(input("Enter the row:"))
y= int(input("Enter the colums:"))
symbol = input("Enter the symbol:")

a=1
while a<=x:
    b=1
    while b<=y:
        print(symbol,end='')
        b+=1
    print()
    a+=1
print("Loop Over")
```
