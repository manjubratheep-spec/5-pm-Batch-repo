### **Python Loop**
### ***While Loop***


```python
a = 1
while a<=10: 
    print(a,end=' ')
    a += 1
print("\nLoop Over")
```

```python
a=10
while a>=0:
    print(a,end=' ')
    a -= 1 
print("\nLoop Over")
```

### **Nested While Loop**

```python
a=1
x= int(input("Enter the row"))
y= int(input("Enter the column"))
while a<=x:
    b=1
    while b<=y:
        print(b,end='')
        b+=1
    print()
    a+=1
print("Loop Over")
```

```python
x= int(input("Enter the row:"))
y= int(input("Enter the colums:"))
symbol = input("Enter the symbol:")

a=1
while a<=x:
    b=1
    while b<=y:
        print(symbol,end='')
        b+=1
    print()
    a+=1
print("Loop Over")
```

### ***For Loop***


```python
mylist = [10,50.7,"hello","world"]
for x in "Hello world":
  print(x,end='')
print()
for y in mylist:
    print(y,end=',')

print("\nLoop Over")
```

```python
for i in range(10+1):
    print(i,end=" ")
print("\n-------------")
for i in range(1,100+1):
    print(i,end=' ')
print("\n-------------")
for i in range(10,100+1,5):
    print(i,end=' ')
print("\n-------------")
for i in range(100,0,-10):
    print(i,end=' ')
print("\nloop over")
```

```python
mylist = [8,9,10,11,15,16,17,20]
for each in mylist:
    if each % 5 == 0:
        print(each,"Divisible By 5")
else:
    print("Number divisible by 5 not found")
```

### **Nested For Loop**

```python
for x in range(5):
    for y in range(1,5+1):
        print(y,end='')
    print()
```

```python
x = int(input("Enter a row :"))
y = int(input("Enter a column :"))
for a in range(x):
    for b in range(1,y+1):
        print(b,end='')
    print()
```

```python
for x in ["black","red","blue"]:
    print("-----------------------------")
    for y in ["Car","bike","dress"]:
        print(x,y)
    print("-----------------------------")
print("Loop Over")
```
