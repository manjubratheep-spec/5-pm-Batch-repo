### **Loop**


```python
a=1
x= int(input("Enter the number"))
y= int(input("Enter the number"))
while a<=x:
    b=1
    while b<=y:
        print(b,end='')
        b+=1
    print()
    a+=1
print("Loop Over")
```

```python
x= int(input("Enter the row:"))
y= int(input("Enter the colums:"))
symbol = input("Enter the symbol:")

a=1
while a<=x:
    b=1
    while b<=y:
        print(symbol,end='')
        b+=1
    print()
    a+=1
print("Loop Over")
```
